Downloading the dataset

```python
import os, subprocess, sys
from pathlib import Path
from dotenv import load_dotenv
```

```python
load_dotenv()  # reads .env from the project root
```

```python
if not os.environ.get("KAGGLE_API_TOKEN"):
    sys.exit("KAGGLE_API_TOKEN not found. Add it to .env (see .env.example).")

RAW = Path.cwd().parent / "data" / "raw" / "paysim.csv"

if not RAW.exists():
    RAW.parent.mkdir(parents=True, exist_ok=True)
    subprocess.run(
        ["kaggle", "datasets", "download", "-d", "ealaxi/paysim1",
         "-p", str(RAW.parent), "--unzip"],
        check=True,
    )

    extracted_csvs = [path for path in RAW.parent.glob("*.csv") if path != RAW]
    if len(extracted_csvs) != 1:
        raise RuntimeError(
            f"Expected exactly one extracted CSV in {RAW.parent}, "
            f"found {len(extracted_csvs)}: {extracted_csvs}"
        )
    extracted_csvs[0].rename(RAW)

print(f"Raw dataset: {RAW.resolve()}")
```
